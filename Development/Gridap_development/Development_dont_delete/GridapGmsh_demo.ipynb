```julia
using Gridap
using GridapGmsh
model = GmshDiscreteModel("GridapGmsh_demo.msh")
labels = get_face_labeling(model)
label_names = model.face_labeling.tag_to_name
```

```julia

order = 1
reffe = ReferenceFE(lagrangian,Float64,order)
# V = TestFESpace(model,reffe,dirichlet_tags=["boundary1","boundary2"])
V = TestFESpace(model,reffe,dirichlet_tags=["Bottom","Top"])
U = TrialFESpace(V,[0,1])
Ω = Triangulation(model)
dΩ = Measure(Ω,2*order)
a(u,v) = ∫( ∇(v)⋅∇(u) )dΩ
l(v) = 0
op = AffineFEOperator(a,l,U,V)
uh = solve(op)
writevtk(Ω,"demo",cellfields=["uh"=>uh])
```
