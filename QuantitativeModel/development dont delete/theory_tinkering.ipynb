```julia
using SymPy
using SpecialFunctions
```

Define symbols and functions

```julia
z, κ₀, α, H, β = symbols("z κ₀ α H β", real=true)
b = symbols("b", cls=SymFunction)
B = symbols("B", cls=SymFunction)
```

Define the ODE (written as == 0 for dsolve)

```julia
ode = Eq( (κ₀ + exp(-α*(z+H))) * diff(b(z), z, 2)
         - α * exp(-α*(z+H)) * diff(b(z),z)
         + B(z), 0)
```

Define the boundary conditions

```julia
# β = 0
bc1 = Eq(diff(b(z),z).subs(z, -H), β)
bc2 = Eq(b(0), 0) ;
```

Homogeneous problem without boundary conditions:

```julia
ode_hom = Eq( (κ₀ + exp(-α*(z+H))) * diff(b(z), z, 2) - α * exp(-α*(z+H)) * diff(b(z),z), 0)
sol = dsolve(ode_hom, b(z))
display(sol)
```

Homogeneous problem with boundary conditions:

```julia
sol = dsolve(ode_hom, b(z), ics = Dict(b(0)=>0, diff(b(z),z).subs(z,-H)=>β) )
display(sol)
```
